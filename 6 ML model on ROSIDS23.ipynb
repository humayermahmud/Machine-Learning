# **Import Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder
from sklearn.metrics import accuracy_score, precision_score,f1_score,recall_score,confusion_matrix,ConfusionMatrixDisplay, roc_auc_score,classification_report

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout


from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
```

# **Load dataset and read**

```python
Dataset_Path = '/content/drive/MyDrive/ML/ROSIDS23.csv'
df = pd.read_csv(Dataset_Path)
```

# **Data Cleaning**

```python
df.info()
df.head()
#df.tail()
#df.shape
```

## **check duplicate rows**

```python

df.duplicated()
#df.duplicated().sum()
#print(df.isnull().sum()[df.isnull().sum() > 0].to_string())# show only the missing value columns with value
df.isnull().sum().sum()
```

## **data types for columns**

```python
#df.dtypes
#df.columns.tolist()
cat_col = df.select_dtypes(include=['object']).columns
print(f"catagorical columns :\n{cat_col}\n")

num_col = df.select_dtypes(include=['number']).columns
print(f"numerical columns :\n{num_col}")
```

```python
zero_variance_cols = num_col[df[num_col].nunique() <= 1]
print("Zero Variance Cols: ", zero_variance_cols.tolist())
```

```python
inf_count = pd.Series(
    np.isinf(df[num_col]).sum(),
    index=num_col
)
print("Infinity Values: \n", inf_count[inf_count > 0])
```

```python
df["Label"].value_counts(normalize=True)
```

```python

df["Label"].value_counts(normalize=True).mul(100).round(2)
```

```python
df.describe().T
```

```python
df_clean = df.drop(columns=zero_variance_cols).copy()

df_clean = df_clean.replace([np.inf, -np.inf], np.nan)

print(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])
```

```python
df_clean = df_clean.dropna()

print("New Shape: ", df_clean.shape)
print("Missing Values: ", df_clean.isnull().sum().sum())
```

# **EDA**

```python

plt.figure(figsize=(8,5))
sns.countplot(data=df, x="Label", order=df["Label"].value_counts().index)

plt.title("Label of Distribution")
plt.xlabel("Label")
plt.ylabel("Count")
plt.xticks(rotation=20)
plt.show()
```

# **Data Preprocessing**

```python
exclude_cols = [
    "Flow ID",
    "Src IP",
    "Dst IP",
    "Timestamp",
    "Label"
]

X = df_clean.drop(columns=exclude_cols)
y = df_clean["Label"]

print("Feature Matrix Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python

print("Features: ", X.count().sum().sum())
```

## **Encoder**

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Label Mapping:")

for i, label in enumerate(label_encoder.classes_):
    print(f"{i} → {label}")
```

## **Train/Test Split**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

print("Training Samples:", X_train.shape[0])
print("Testing Samples:", X_test.shape[0])
```

## **Scaling**

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_train_scaled.std()
```

# **Machine Learning Model**

## ***01. Random Forest***

```python
rf = RandomForestClassifier(
    n_estimators = 100,
    random_state = 42,
    n_jobs = -1
)
rf.fit(X_train, y_train)
```

```python
rf_pred = rf.predict(X_test)
rf_prob =rf.predict_proba(X_test)
```

```python
print("Accuracy Score: ", accuracy_score(y_test, rf_pred))
print("Precision Score: ", precision_score(y_test, rf_pred, average="macro"))
print("Recall Score: ", recall_score(y_test, rf_pred, average="macro"))
print("F1 Score: ", f1_score(y_test, rf_pred, average="macro"))
print("Roc Auc Score: ", roc_auc_score(y_test, rf_prob, multi_class='ovr', average='macro'))
print("\n")
print(classification_report(y_test, rf_pred))
print("\n")
print(confusion_matrix(y_test, rf_pred))
```

```python
plt.figure(figsize=(10,6))

plt.title("Random Forest Model Evaluation Metrics")
plt.xlabel("Metrics")
plt.ylabel("Values")

matrics = ["Accuracy", "Precision", "Recall", "F1 Score", "Roc Auc Score"]
values = [
    accuracy_score(y_test, rf_pred),
    precision_score(y_test, rf_pred, average="macro"),
    recall_score(y_test, rf_pred, average="macro"),
    f1_score(y_test, rf_pred, average="macro"),
    roc_auc_score(y_test, rf_prob, multi_class='ovr', average='macro')
]

bars = plt.bar(matrics, values)
plt.bar_label(bars, fmt='%.2f')
plt.show()
```

# ***Comparision of All ML Model***

```python
model_names = [
    "Random Forest",
    "Decision Tree",
    "K-Nearest Neighbors",
    "Support Vector Classification",
     "XGBoost",
    "LightGBM"
]

accuracy_scores = [
    accuracy_score(y_test, rf_pred),
    accuracy_score(y_test, dt_pred),
    accuracy_score(y_test, knn_pred),
    accuracy_score(y_test, svm_pred),
    accuracy_score(y_test, xgb_pred),
    accuracy_score(y_test, lgbm_pred)
]

precision_scores = [
    precision_score(y_test, rf_pred, average="macro"),
    precision_score(y_test, dt_pred, average="macro"),
    precision_score(y_test, knn_pred, average="macro"),
    precision_score(y_test, svm_pred, average="macro"),
    precision_score(y_test, xgb_pred, average="macro"),
    precision_score(y_test, lgbm_pred, average="macro"),
]

recall_scores = [
    recall_score(y_test, rf_pred, average="macro"),
    recall_score(y_test, dt_pred, average="macro"),
    recall_score(y_test, knn_pred, average="macro"),
    recall_score(y_test, svm_pred, average="macro"),
    recall_score(y_test, xgb_pred, average="macro"),
    recall_score(y_test, lgbm_pred, average="macro"),
]

f1_scores = [
    f1_score(y_test, rf_pred, average="macro"),
    f1_score(y_test, dt_pred, average="macro"),
    f1_score(y_test, knn_pred, average="macro"),
    f1_score(y_test, svm_pred, average="macro"),
    f1_score(y_test, xgb_pred, average="macro"),
    f1_score(y_test, lgbm_pred, average="macro"),
]

roc_auc_scores = [
    roc_auc_score(y_test, rf_prob, multi_class="ovr", average="macro"),
    roc_auc_score(y_test, dt_prob, multi_class="ovr", average="macro"),
    roc_auc_score(y_test, knn_prob, multi_class="ovr", average="macro"),
    roc_auc_score(y_test, svm_prob, multi_class="ovr", average="macro"),
    roc_auc_score(y_test, xgb_prob, multi_class="ovr", average="macro"),
    roc_auc_score(y_test, lgbm_prob, multi_class="ovr", average="macro"),
]


comparison_df = pd.DataFrame({
    "Model": model_names,
    "Accuracy": accuracy_scores,
    "Precision": precision_scores,
    "Recall": recall_scores,
    "F1 Score": f1_scores,
    "ROC AUC Score": roc_auc_scores
})

display(comparison_df.set_index('Model'))
```
